### 28. Logical and Comparison Operators

Common operators used in R:

- `>` Greater than
- `>=` Greater than or equal to
- `<` Less than
- `<=` Less than or equal to
- `==` Exactly equal to
- `!=` Not equal to
- `!` Logical NOT
- `&`, `&&` Logical AND
- `|`, `||` Logical OR
- `xor(x, y)` Exclusive OR
- `isTRUE(x)` Tests whether `x` is TRUE
- `isFALSE(x)` Tests whether `x` is FALSE


### 29. Example of `|` and `||`

```r
x <- 8
(x < 10) || (x < 2)

x <- 18
(x < 10) || (x < 2)
```

### 30. Element-wise `|` versus `||`

```r
x <- c(8, 18)

(x < 10) || (x < 2)
(x < 10) | (x < 2)
```

### 31. Example of `&` and `&&`

```r
x <- 5
(x < 10) && (x > 2)

x <- 15
(x < 10) && (x > 2)
```

### 32. Element-wise `&` versus `&&`

```r
x <- c(8, 18)

(x < 10) && (x > 2)
(x < 10) & (x > 2)
```

### 33. Filtering a Vector using Logical Conditions

```r
x <- 1:6 # 1,2,3,4,5,6

(x > 2) & (x < 5)

x[(x > 2) & (x < 5)]
```

### 34. Filtering using Logical OR

```r
x <- 1:6

(x > 2) | (x < 5)

x[(x > 2) | (x < 5)]
```

### 35. Understanding `&&` with a Vector

```r
x <- 1:6

(x > 2) && (x < 5)

# Equivalent to checking only the first element
(x[1] > 2) & (x[1] < 5)
```

### 36. Relational Operators

`TRUE` and `FALSE` are logical constants in R.

Also, in logical contexts, zero is treated as FALSE and non-zero numbers are treated as TRUE.

### 38. Checking Logical Results with `is.logical()`

```r
x <- 5

Logical1 <- (x > 2)
Logical1
is.logical(Logical1)

Logical2 <- (x < 10)
Logical2
is.logical(Logical2)

Logical3 <- (x != 5)
Logical3
is.logical(Logical3)
```

### 39. More Logical Expressions

```r
x <- 5

Logical4 <- (2 * x > 11)
Logical4
is.logical(Logical4)

Logical5 <- (3 * x < 20)
Logical5
is.logical(Logical5)
```
