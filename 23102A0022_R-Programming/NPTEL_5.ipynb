### Conditional execution: if else()

```r
x=5
if(x==3)
{
  x=x-1
}else
{
  x=2*x
}
x
```

```r
x=2
if(x==2) {x=x-1} else {x=2*x}
x
```

```r
x=6
if(x>3)
{
  print("The value is more than 3")
} else {" The value is not more than 3"}
```
