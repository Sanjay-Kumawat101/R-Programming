# **Strings - Display and Formatting**

The print function has a significant limitation that it prints only one
object at a time.

```r
print("The zero occurance at", 2*pi, "radians")
```

The only way to print multiple items is to print them one at a time

```r
print("The zero occurs at");
print(2*pi);
print("radians")
```

**Formatting and Display of Strings: cat function**

The cat function is an alternative to print that lets you combine
multiple items into a continuous output.

```r
print("The zero occurs at");
print("2*pi");
print("radians")
```

```r
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

```r
d = date()
cat("Today's date is:", d, "\n")
```

```r
x = 1:10
x

cat(x, sep =" ++ ")
cat("\n")

cat(x, sep=" /")
```

```r
x = 7
cat("The square of", x, "is", x^2, "!\n")
```

```r
cat("The square root of", x, "is approximtely", format(sqrt(x), digits=3), "\n")
```

```r
evenno = c(2,4,6,8,10)
evenno

cat("The first few even numbers are:", evenno, "....\n")
```

```r
x = 1:10
x

cat(x, fill = 2, labels = paste("(", letters[1:10], "):"))
```

**paste() function**

```r
paste(1:12)

as.character(1:12)
```

```r
paste("Everday", "loves", "R Programming.")
paste("Everday", "loves", "R Programming.", sep="*")
paste("Everday", "loves", "R Programming.", sep="===")
```

```r
names = c("Prof. Singh", "Mr. Venkat", "Dr. Jha")
names

paste(names, "is", "a good", "person.")
```

```r
names = c("Prof. Singh", "Mr. Venkat", "Dr. Jha")
names

paste(names, "is", "a good", "person.", collapse=", and ")
```

**Operations with Strings**

```r
paste(1, " is first", 2, " is second", 3, " is third", sep = "#")
paste(1, " is first   ", 2, " is second   ", 3, " is third    ", sep = "#")
```

```r
x = paste("Ex", 1:5, sep ="_")
x

x[1]
x[2]
x[3]
x[5]
```

```r
x = paste("Ex", 1:5, sep="_", collapse="")
x
```

```r
x = paste("Ex", 1:5, sep="_")
x
```

```r
paste0(1:10)
paste(1:10)
```

```r
paste0(1:10, c("st", "nd", "rd", rep("th", 7) ))
paste(1:10, c("st", "nd", "rd", rep("th", 7) ))
```

**String Splitting**

```r
x = "The&!syntax&!of&!paste&!is!&available!
&inthe online-help"
x
strsplit(x, split="!")
```

```r
x = "The&!syntax&!of&!paste&!is!&available!
&inthe online-help"
x
strsplit(x, split="&!")
```

```r
x = "The&!syntax&!of&!paste&!is!&available!
&inthe online-help"
x

y = strsplit(x, split = "!&")
y

y[[1]][1]
y[[1]][2]
y[[1]][3]
```

```r
dates = c("2020-07-24", "2021-08-25", "2022-
09-26", "2023-10-27")

datesplt = strsplit(dates, "-")
datesplt
```

**Convert to matrix**

```r
datemat = matrix(unlist(datesplt), nrow = 4,
ncol=3, byrow=TRUE)

datemat
```

```r
datematrix =
matrix(as.numeric(unlist(datesplt)), nrow = 4,
ncol=3, byrow=TRUE)

datematrix
```

**Manipulations with Strings and Alphabets**

```r
x = "R course 24.09.2026"
x

y = "Number of participants: 4460"
y

nchar(x)
nchar(y)
nzchar(x)
nzchar(y)
```

```r
x = c("Apple", "Banana", "Cake")
x

nzchar(x)

y = c("Apple", "", "Cake")
y

nzchar(y)
```

```r
x = c("Apple", "Banana", "Cake")
nchar(x)

y =c(2,4,6)
nchar(y)

z = c(11,222,3333)
nchar(z)

z1 = c(1.1, 2.22, 3.333)
nchar(z1)
```
