1. The for loop

```r
for(i in 1:5)
{
    print(i ^ 2)
}
```

```r
for(i in c(2, 4, 6, 7))
{
  print(i ^ 2)
}
```

```r
x = c(2, 4, 6, 8, 10, 12)

excount = function(x)
{
  count = 0

  for(xval in x)
  {
    if(xval / 2 > 3)
    count = count+1
  }
  print(count)
}
excount(x)
```

2. Nested looping with for loop

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for(x in child)
{
  for(y in sweet)
  {
    print(paste(x, y))
  }
}
```

3. The break command

Using the break command, we can stop the loop before it has
looped through all the items.

```r
drink = c("coffee", "lemonade", "tea", "juice")

for(x in drink)
{
  if(x == "tea")
  {
    break
  }
  print(x)
}
```

4. The next command

Using the next command, we can skip an iteration without
terminating the loop.

```r
drink = c("coffee", "lemonade", "tea", "juice")

for(x in drink)
{
  if(x == "lemonade")
  {
    next
  }
  print(x)
}
```

```r
drink = c("coffee", "lemonade", "tea", "juice")

for(x in drink)
{
  if(x == "tea")
  {
    next
  }
  print(x)
}
```

5. The while loop

```r
i = 1

while(i < 10)
{
  print(i ^ 2)
  i = i + 2
}
```

```r
sumfunction  = function()
{
  sum = 0
  number = as.integer(readline(prompt="Please select any number less than 25: "))

  while(number <= 25)
  {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While Loop: ", sum))
}
sumfunction()
```

6. The repeat loop

```r
i = 1

repeat
{
  print(i ^ 2)
  i = i + 2

  if(i > 10)
  break
}
```

```r
i = 1

repeat
{
  i = i + 1
  if(i < 10)
  next

  print(i ^ 2)
  if(i >= 13)
  break
}
```

7. Functions (Single Variable)

```r
abc = function(x)
{
  x^2
}

abc(3)
abc(6)
abc(9)
```

8. Functions (Two Variables)

```r
abc = function(x, y)
{
  x^2+y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

```r
abc = function(x)
{
  sin(x)^2+cos(x)^2+x
}

abc(9)
abc(99)
abc(-15)
```

```r
abc = function()
{
  for(i in 1:3)
  {
    print(i ^ 3)
  }
}
abc()
```
